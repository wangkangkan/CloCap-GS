```python
# 获取服装边缘边
import trimesh
import numpy as np

from collections import defaultdict

def get_unique_edges(vertices, faces):
    # 用于存储每条边出现的次数
    edge_count = defaultdict(int)
    
    # 遍历每个面，计算每条边的出现次数
    for face in faces:
        for i in range(3):
            edge = tuple(sorted([face[i], face[(i + 1) % 3]]))  # 保证边的顺序一致性
            edge_count[edge] += 1
    
    # 筛选出只出现一次的边（即只属于一个面的边）
    unique_edges = [edge for edge, count in edge_count.items() if count == 1]
    
    return np.array(list(edge)).astype(np.int32), np.array(unique_edges).astype(np.int32)

# NOTE: 设置服装路径
mesh = trimesh.load("../data/Franzired/FranziRed3550/templatedeformT/cloth/clothes.obj")
vertices = mesh.vertices
faces = mesh.faces
# 获取只属于一个面的边
edges, unique_edges = get_unique_edges(vertices, faces)
print("Unique Edges (belong to only one face):", unique_edges)
print("Unique Edges nums:", len(unique_edges))
```

```python
# 找到拥有共同点的边
def get_edges_with_common_vertex(edges, faces):
    # 用于存储每个顶点对应的边集合
    vertex_edges = defaultdict(list)
    
    # 遍历每个面，计算边与顶点的关系
    for i, face in enumerate(faces):
        for j in range(3):
            edge = tuple(sorted([face[j], face[(j + 1) % 3]]))  # 保证边的顺序一致性
            vertex_edges[face[j]].append(edge)
            vertex_edges[face[(j + 1) % 3]].append(edge)
    
    return vertex_edges

# 找到边缘边具有共同顶点的边对
def find_neighbor_edges(unique_edges):
    vert_set = set()
    edges_pair = []
    
    for i,edge in enumerate(unique_edges):
        vert_set.add(edge[0])
        vert_set.add(edge[1])
        
    for v in vert_set:
        for i,edge in enumerate(unique_edges):
            if edge[0] == v or edge[1] == v:
                for j in range(i+1, len(unique_edges)):
                    if unique_edges[j][0] == v or unique_edges[j][1] == v:
                        e1 = [edge[0], edge[1]] if edge[0] == v else [edge[1], edge[0]]
                        e2 = [unique_edges[j][1], unique_edges[j][0]] if unique_edges[j][0] == v else [unique_edges[j][0], unique_edges[j][1]]
                        edges_pair.append([e1, e2])
    
    return np.array(edges_pair).astype(np.int32)

edges_pair = find_neighbor_edges(unique_edges)
print(edges_pair.shape)
# print(edges_pair)
# print("edges_pair num:", len(edges_pair))

vertex_colors = np.ones((len(vertices), 4))
# idx = np.random.randint(0, len(edges_pair))
# edges1 = edges_pair[0, 0]
# edges2 = edges_pair[0, 1]
# print(edges1)

# v1 = unique_edges[edges1, 0]
# v2 = unique_edges[edges1, 1]
# v3 = unique_edges[edges2, 0]
# v4 = unique_edges[edges2, 1]
e1  = edges_pair[:, 0]
print(e1)
e2 = edges_pair[:, 1]
print(e2)
v1 = e1[:, 0]
v2 = e1[:, 1]
v3 = e2[:, 1]
v4 = e2[:, 0]
vertex_colors[v1] = np.array([255.0, 0.0, 0.0, 1.0])
vertex_colors[v2] = np.array([255.0, 0.0, 0.0, 1.0])
vertex_colors[v3] = np.array([255.0, 0.0, 0.0, 1.0])
vertex_colors[v4] = np.array([255.0, 0.0, 0.0, 1.0])

mesh.visual.vertex_colors = vertex_colors

print("v1:", v1, "v2:", v2, "v3:",v3, "v4:", v4)
# mesh.show()
# mesh.export("test.obj")

np.save("edges_pair.npy", edges_pair)
```
