```python
# 测试pytorch3d细分网格
import numpy as np
import torch
import trimesh
from pytorch3d.ops import subdivide_meshes
from pytorch3d.structures import Meshes

device = "cuda"
# mesh = trimesh.load("../tmp/deformcloth.obj")
mesh = trimesh.load("../data/magdalena/magdalena2000-allviews/templatedeformT/cloth/clothes.obj")
vertices = torch.from_numpy(mesh.vertices)
faces = torch.from_numpy(mesh.faces)
verts_features = torch.zeros([1, vertices.shape[0], 3])
surface_mesh = Meshes(
    verts=[vertices.to(device)],   
    faces=[faces.to(device)],
    #textures=TexturesVertex(verts_features=verts_features.clamp(0, 1).to(device)),
    # verts_normals=[verts_normals.to(rc.device)],
)

submesh = subdivide_meshes.SubdivideMeshes()
mesh = submesh(surface_mesh)
new_faces = mesh.faces_list()[0].cpu().numpy()
np.save("../data/magdalena/magdalena2000-allviews/templatedeformT/cloth/faces_more.npy", new_faces)
new_verts = mesh.verts_list()[0].cpu().numpy()
np.save("../data/magdalena/magdalena2000-allviews/templatedeformT/cloth/verts_more.npy", new_verts)
```
